## **Expressions:**

An expression is a combination of values, variables, and operators that can be evaluated to produce a single value.

Expressions can be simple, involving only values and operators, or more complex, involving function calls and more advanced operations.

````python
x = 5
y = x + 10  # 'x + 10' is an expression; it involves the variable 'x', a value (10), and an operator (+)
````

## **Basic Operators in Python**
Python provides a variety of operators, which are the building blocks of most expressions:
* Arithmetic operators,
* comparison operators,
* Logical operators,
* Assignment operators,
* Bitwise Operators:


**Arithmetic Operators:**

* Addition :  +
* Subtraction: -
*  Multiplication: *
*  Division: /
*  Modulus (remainder) : %
* Exponentiation: **
*  Floor division: //

````python
a = 10
b = 3
print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.3333...
print(a % b)   # 1
print(a ** b)  # 1000
print(a // b)  # 3
````

**Comparison Operators (return True or False):**

* Equal to :  ==
*  Not equal to: !=
* Less than :  <  
* Greater than:  >
* Less than or equal to:  <=
*  Greater than or equal to:  >=

````python
x = 5
y = 10
print(x == y)  # False
print(x != y)  # True
print(x < y)   # True
print(x > y)   # False
print(x <= y)  # True
print(x >= y)  # False
````

**Assignment Operators:**

**Bitwise Operators:**

* Bitwise AND : &
* Bitwise OR: |
*  Bitwise XOR: ^
*  Bitwise NOT: ~

````python
a = 60  # 60 = 0011 1100 in binary
b = 13  # 13 = 0000 1101 in binary
print(a & b)   # 12 = 0000 1100
print(a | b)   # 61 = 0011 1101
print(a ^ b)   # 49 = 0011 0001
print(~a)      # -61 (in two's complement form)
print(a << 2)  # 240 = 1111 0000
print(a >> 2)  # 15 = 0000 1111
````

**Problem:** Calculate the Total Cost of a Road Trip

**Scenario:** You are planning a road trip. To budget for the trip, you want to calculate the total cost of fuel for the entire journey. You have the following information:

The total distance of the trip in miles.
The fuel efficiency of your car in miles per gallon.
The cost of fuel per gallon.

**Objective:**
 Calculate the total cost of fuel for the trip.*

````python
# Variables provided for the scenario

# Total distance of the trip in miles
distance = 500  # e.g., 500 miles

# Fuel efficiency of the car (miles per gallon)
fuel_efficiency = 25  # e.g., 25 mpg. This means the car can travel 25 miles with 1 gallon of fuel.

# Cost of fuel per gallon
cost_per_gallon = 3.50  # e.g., $3.50 per gallon

# Using expressions to calculate the total fuel required and the total cost

# Total fuel required for the trip (distance / fuel efficiency)
total_fuel = distance / fuel_efficiency  # Using the division operator to determine gallons of fuel required

# Total cost of fuel for the trip (total fuel * cost per gallon)
total_cost = total_fuel * cost_per_gallon  # Using the multiplication operator to calculate the cost

# Displaying the result
print(f"The total cost of fuel for the trip is ${total_cost:.2f}")  # Using an f-string to format the result

# Output: The total cost of fuel for the trip is $70.00
````

## Conditional Statements

Conditional statements in programming are crucial because they allow a program to make decisions and execute different code paths based on certain conditions.

Just like humans make decisions every single day on what cloths to wear depending on the weather conditions, when the cross the street based on traffic light color, attend this python bootcamp depending on wether you ate breakfast or not, etc.


In Python, the primary conditional statements are introduced using if, elif, and else

 Basically conditional statements tests whether a condition is True or False.

**Why Conditional Statements?**

**Decision Making:** Just as in real life, where we make decisions based on certain conditions (e.g., if it's raining, we'll take an umbrella), programs often need to decide what action to take based on particular circumstances. Conditional statements allow this decision-making process.

**Dynamic Behavior:** They allow the program to adapt and produce different outcomes depending on input or other factors. Without conditionals, programs would always follow the same path and provide the same output.

**Error Handling:** Conditional checks can help a program identify unexpected situations or errors and handle them appropriately, preventing crashes or incorrect outcomes.

**Optimization:** By using conditional statements, a program can choose the most efficient way to perform a task based on the current conditions.

**Comparison Operators**
In Python, comparison operators are used to compare values, and they return either True or False. These operators play a central role in conditional statements, specifically in the if statement, as they determine whether a particular block of code should be executed.

The if, elif, and else statements in Python are pivotal for decision-making processes. They allow for sequential checks and execute specific code blocks based on conditions.


* **if statement:** Starts with the if keyword, followed by a condition and a colon (:). The code block to execute if the condition is True is indented below the if.

* **elif statement:** Short for "else if". It allows for additional conditions after the initial if. It comes after an if or another elif, followed by a condition and a colon (:).

* **else statement:** This doesn't have a condition. It catches anything that wasn't caught by the preceding if or elif statements.

**Basic Structure**

````python
if condition1:
    # code block for condition1
elif condition2:
    # code block for condition2
else:
    # code block if none of the conditions are met
````

**Example 1: A simple if-else statement.**

````python
age = 17
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
````

**Scenario:** A movie theater has the following ticket pricing policy:

* Kids under the age of 10 pay \$5.
* Teenagers between the ages of 10 and 17 pay \$7.
* Adults between the ages of 18 and 59 pay \$10.
* Senior citizens, 60 years or older, pay \$8.

On Wednesdays, everyone gets a $2 discount, regardless of their age group.
Given the age of the person and the day of the week, determine the ticket price.

````python
# Input age and day of the week
age = int(input("Enter your age: "))
day = input("Enter the day of the week: ").lower()

# Determine the base ticket price
if age < 10:
    price = 5
elif age <= 17:
    price = 7
elif age <= 59:
    price = 10
else:
    price = 8

# Apply discount if it's Wednesday
if day == "wednesday":
    price -= 2

# Display the final ticket price
print(f"The ticket price is ${price}.")
````
